```python
%sql
CREATE TABLE IF NOT EXISTS voltgrid.source.meta_trigger_data (
  file_id          BIGINT GENERATED ALWAYS AS IDENTITY,
  entity_name      STRING     NOT NULL COMMENT 'customers, payment, sessions, stations, vehicles',
  file_path        STRING     NOT NULL COMMENT 'folder path, e.g. raw/customers/ingestion_date=2026-10-09',
  file_name        STRING     NOT NULL,
  file_size_bytes  BIGINT,
  process_flag     STRING     NOT NULL COMMENT 'N = new, P = processing, Y = processed, F = failed',
  arrived_at       TIMESTAMP  NOT NULL,
  processed_at     TIMESTAMP,
  pipeline_run_id  STRING,
  error_message    STRING
)
USING DELTA
COMMENT 'One row per arrived file; process_flag drives pipeline triggering';
```
