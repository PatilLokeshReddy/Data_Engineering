```sql
create table if not exists voltgrid.gold.total_revenue as(
    select sum(p.amount_aud) over(partition by st.state_code) as total_revenue_as_per_state,st.state_code
    from voltgrid.silver.sessions s 
    join voltgrid.silver.customers c on c.customer_id=s.customer_id
    join voltgrid.silver.payments p on p.customer_id=c.customer_id
    join voltgrid.silver.stations st on st.station_id=s.station_id
)
```
