```sql
create table if not exists voltgrid.gold.total_revenue_by_cust as(
    select sum(p.amount_aud+p.gst) over(partition by c.customer_id) as total_revenue,sum(p.amount_aud+p.gst) over(partition by c.customer_id order by p.processed_at ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW) cust_running_total,p.payment_mode,c.full_name,c.email,c.phone,c.loyalty_tier
    from voltgrid.silver.customers c join voltgrid.silver.payments p on c.customer_id=p.customer_id
    where c.active_flg='Y'
)
```
